```python
import matplotlib.pyplot as plt

df["Attrition"].value_counts().plot(
    kind="bar",
    title="Employee Attrition",
    xlabel="Attrition",
    ylabel="Number of Employees"
)

plt.tight_layout()
plt.savefig("../dashboard/screenshots/attrition_overview.png", dpi=300)
plt.show()
```

```python
department_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"],
    normalize="index"
) * 100

department_attrition["Yes"].sort_values().plot(
    kind="bar",
    title="Observed Attrition Rate by Department",
    xlabel="Department",
    ylabel="Attrition Rate (%)"
)

plt.tight_layout()
plt.savefig("../dashboard/screenshots/attrition_by_department.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt

overtime_attrition = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
) * 100

overtime_attrition["Yes"].sort_values().plot(
    kind="bar",
    title="Observed Attrition Rate by Overtime",
    xlabel="Overtime",
    ylabel="Attrition Rate (%)"
)

plt.tight_layout()
plt.savefig("../dashboard/screenshots/attrition_by_overtime.png", dpi=300)
plt.show()
```

```python
jobrole_attrition = pd.crosstab(
    df["JobRole"],
    df["Attrition"],
    normalize="index"
) * 100

jobrole_attrition["Yes"].sort_values().plot(
    kind="barh",
    title="Observed Attrition Rate by Job Role",
    xlabel="Attrition Rate (%)",
    ylabel="Job Role"
)

plt.tight_layout()
plt.savefig("../dashboard/screenshots/attrition_by_job_role.png", dpi=300)
plt.show()
```

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 24, 34, 44, 54, 100],
    labels=["Under 25", "25-34", "35-44", "45-54", "55+"]
)

age_attrition = pd.crosstab(
    df["AgeGroup"],
    df["Attrition"],
    normalize="index"
) * 100

age_attrition["Yes"].plot(
    kind="bar",
    title="Observed Attrition Rate by Age Group",
    xlabel="Age Group",
    ylabel="Attrition Rate (%)"
)

plt.tight_layout()
plt.savefig("../dashboard/screenshots/attrition_by_age_group.png", dpi=300)
plt.show()
```

```python
joblevel_attrition = pd.crosstab(
    df["JobLevel"],
    df["Attrition"],
    normalize="index"
) * 100

joblevel_attrition["Yes"].plot(
    kind="bar",
    title="Observed Attrition Rate by Job Level",
    xlabel="Job Level",
    ylabel="Attrition Rate (%)"
)

plt.tight_layout()
plt.savefig("../dashboard/screenshots/attrition_by_job_level.png", dpi=300)
plt.show()
```

```python
df["TenureGroup"] = pd.cut(
    df["YearsAtCompany"],
    bins=[-1, 1, 3, 5, 10, 100],
    labels=["0-1", "2-3", "4-5", "6-10", "11+"]
)

tenure_attrition = pd.crosstab(
    df["TenureGroup"],
    df["Attrition"],
    normalize="index"
) * 100

tenure_attrition["Yes"].plot(
    kind="bar",
    title="Observed Attrition Rate by Years at Company",
    xlabel="Years at Company",
    ylabel="Attrition Rate (%)"
)

plt.tight_layout()
plt.savefig("../dashboard/screenshots/attrition_by_tenure.png", dpi=300)
plt.show()
```
